```python
import base64, json, zlib
from IPython.display import SVG, display
import subprocess

graph = 'flowchart TD\n    A([Начало]) --> B[/Ввести целое число n/]\n    B --> C{n mod 2 = 0?}\n    C -->|Да| D[/Вывести «Чётное»/]\n    C -->|Нет| E[/Вывести «Нечётное»/]\n    D --> F([Конец])\n    E --> F'
state = {"code": graph, "mermaid": {"theme": "default"}}
packed = zlib.compress(json.dumps(state, ensure_ascii=False).encode("utf-8"))
token = "pako:" + base64.urlsafe_b64encode(packed).decode("ascii").rstrip("=")
url = "https://mermaid.ink/svg/" + token
print(url)
svg = subprocess.check_output(["curl", "-fLsS", "--max-time", "30", url]).decode("utf-8")
display(SVG(data=svg))
```

```python
def parity(n):
    return "Чётное" if n % 2 == 0 else "Нечётное"

cases = [(-3, "Нечётное"), (0, "Чётное"), (7, "Нечётное"), (12, "Чётное")]
for n, expected in cases:
    actual = parity(n)
    assert actual == expected
    print(f"n = {n:3}: {actual} — верно")
print("Проверки: 4 из 4 выполнены успешно")
```
